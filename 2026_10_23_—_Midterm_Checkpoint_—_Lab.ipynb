```python
# Standardize text case to uppercase and strip

def to_uppercase(column_name):
  standardized = df[column_name].str.strip().str.upper()
  rows_affected = df[column_name].ne(standardized).sum() # Count number of rows affected
  df[column_name] = standardized

  return df, rows_affected

# Standardize store_id values by stripping leading/trailing whitespace and converting all characters to uppercase.
# This ensures consistent formatting for matching store IDs when joining with the store dataset.
# EX: FL-710 vs. fl-412 - We want all to be FL-710

df, rows_affected = to_uppercase('store_id')

log(
    'Data cleaning',
    'store_id: standardize text case to uppercase and strip',
    rows_affected
)

df.head()
```

```python
# Standardize text case to title case and strip

def to_title_case(column_name):
  standardized = df[column_name].str.strip().str.title()
  rows_affected = df[column_name].ne(standardized).sum() # Count number of rows affected
  df[column_name] = standardized

  return df, rows_affected

# Standardize product_name and category column information
# EX: PEANUT BUTTER - We want it to be Peanut Butter

df, rows_affected = to_title_case('product_name')
log('Data cleaning', 'product_name: standardize text case to title case and strip', rows_affected)

df, rows_affected = to_title_case('category')

log(
    'Data cleaning',
    'category: standardize text case to title case and strip',
    rows_affected
)

df.head()
```

```python
# Look at the datatypes of the columns

print('Column datatypes:\n', df.dtypes)
```

```python
# Update unit_price to be float type, remove $, and strip

def to_float(column_name):
  standardized = df[column_name].str.strip().str.replace('$', '', regex=False).astype(float)
  rows_affected = df[column_name].ne(standardized).sum() # Count number of rows affected
  df[column_name] = standardized

  return df, rows_affected

# unit_price is a String but it should be a float type for future calculations (i.e. summing/etc)
df, rows_affected = to_float('unit_price')

log(
    'Data cleaning',
    'unit_price: update to be float type and remove $',
    rows_affected
)

df.head()
```

```python
# Post-function analysis

print('Post-to_float()\n')
print('Datatype:', df['unit_price'].dtype)
```

```python
# Duplicate rows function

def drop_duplicate_rows(df):
  rows_affected = df.duplicated().sum() # Count number of rows affected
  df.drop_duplicates(inplace=True)

  return df, rows_affected

# Remove exact duplicate rows because they represent repeated records that could inflate transaction counts and sales totals
# Only rows identical across all columns are removed
# EX: rows that share a transaction_id are not automatically considered duplicates

df, rows_affected = drop_duplicate_rows(df)

log(
    'Data cleaning',
    'Drop duplicate rows',
    rows_affected
)

df.head()
```

```python
# Post-function analysis

print('Post-to_float()\n')
print('Number of rows:', df.shape[0])
```

```python
# Negative quantity value function

def change_negative_quantities(column_name):
  rows_affected = (df[column_name] < 0).sum() # Count number of rows affected
  df[column_name] = df[column_name].abs()

  return df, rows_affected

# Convert negative quantity values to positive values using abs() so all quantities represent nonnegative units sold
# We decided to make the quantity values positive because negative quantities do not make sense
# Instead of dropping the rows, we decided to view negative quantities as user entry mistakes
df, rows_affected = change_negative_quantities('quantity')

log(
    'Data cleaning',
    'quantity: Changed negative quantity values to positive values',
    rows_affected
)
```

```python
# Pre-function analysis

print('Pre-parse_timestamp()\n')
print('Unique values', df['timestamp'].unique())
print('Has NaN values:', df['timestamp'].hasnans)
print('NaN count:', df['timestamp'].isna().sum())
```

Originally, we had a standardized = pd.to_datetime(df[column_name], errors='coerce'). errors='coerce' tells Pandas to convert valid timestamp strings into datetime objects, convert missing values (NaN, None, etc) into NaT, and convert unparseable timestamp values into NaT. This 3rd case is what made our NA value count go from 884 to 23028 post-function application, meaning that there were many cases where the original timestamp values were not NaN but were converted to NaT. In other words, many of the original timestamp values may not be compatible with Pandas' datetime parsing rules.

To better handle these cases, we decided to use a format='mixed' parameter. This lets Pandas infer the format of each value individually, which is useful because our dataset contains multiple timestamp formats.

Furthermore, to maintain the original timestamp values for comparison, we created a copy of the column using .copy() before parsing. This allowed us to distinguish between values that were already missing and values that failed to parse during conversion. We then updated the rows_affected calculation to count only values that were originally non-missing but became NaT after parsing, giving us a more accurate measure of parsing failures.

```python
# Parse timestamp function by converting to date time format

def parse_timestamp(column_name):
  original = df[column_name].copy()

  standardized = pd.to_datetime(original, errors='coerce', format='mixed')
  rows_affected = (original.notna() & standardized.isna()).sum()
  df[column_name] = standardized

  return df, rows_affected

df, rows_affected = parse_timestamp('timestamp')

log(
    'Data cleaning',
    'timestamp: Parse timestamp function by converting to date time format',
    rows_affected
)
```

**Clean product_name**

To figure out which product_name we wanted to use, we decided to look up the various product names online to see what the most standardized way of addressing these columns were. For instance, for the beer, there were two different ways of saying it - "Light Beer" and "Beer Light". "Light Beer" seemed to be the more standardized way of addressing this product. Moreover, we put descriptors of size, pack size, and fluid size in parenthesis.

```python
# Find product names that contain a keyword, ignoring capitalization and missing values

# Returns the SKUs, names, and categories related to the description
def clean_product_name(description):
  product = df.loc[df['product_name'].str.lower().str.contains(description.lower(), na=False)]

  # Number of rows
  # print(f'There are {len(product)} rows that look like {description} products.\n')

  # List SKUs, names, and categories
  return product[['sku', 'product_name', 'category']].drop_duplicates()
```
