### 1 — Your cleaning pipeline as functions

**TODO:** paste the cleaning code you have written so far. It should be functions, not loose cells, and it should cover at minimum: duplicates, `unit_price` to numeric (it comes in as text, some with a `$`), the negative `quantity` values with your decision applied, `timestamp` parsed, and `store_id` cleaned up so it will join to the store file.

Print the row count before and after. We want to see how many rows your decisions removed.

```python
# Standardize text case to uppercase and strip

def to_uppercase(column_name):
  standardized = df[column_name].str.strip().str.upper()
  rows_affected = df[column_name].ne(standardized).sum() # Count number of rows affected
  df[column_name] = standardized

  return df, rows_affected

# Standardize store_id values by stripping leading/trailing whitespace and converting all characters to uppercase.
# This ensures consistent formatting for matching store IDs when joining with the store dataset.
# EX: FL-710 vs. fl-412 - We want all to be FL-710

df, rows_affected = to_uppercase('store_id')

log(
    'Data cleaning',
    'store_id: standardize text case to uppercase and strip',
    rows_affected
)

df.head()
```

```python
# Standardize text case to title case and strip

def to_title_case(column_name):
  standardized = df[column_name].str.strip().str.title()
  rows_affected = df[column_name].ne(standardized).sum() # Count number of rows affected
  df[column_name] = standardized

  return df, rows_affected

# Standardize product_name and category column information
# EX: PEANUT BUTTER - We want it to be Peanut Butter

df, rows_affected = to_title_case('product_name')
log('Data cleaning', 'product_name: standardize text case to title case and strip', rows_affected)

df, rows_affected = to_title_case('category')

log(
    'Data cleaning',
    'category: standardize text case to title case and strip',
    rows_affected
)

df.head()
```

```python
# Look at the datatypes of the columns
df.dtypes
```

```python
# Update unit_price to be float type, remove $, and strip

def to_float(column_name):
  standardized = df[column_name].str.strip().str.replace('$', '', regex=False).astype(float)
  rows_affected = df[column_name].ne(standardized).sum() # Count number of rows affected
  df[column_name] = standardized

  return df, rows_affected

# unit_price is a String but it should be a float type for future calculations (i.e. summing/etc)
df, rows_affected = to_float('unit_price')

log(
    'Data cleaning',
    'unit_price: update to be float type and remove $',
    rows_affected
)

df.head()
```

```python
df['unit_price'].dtype
```

```python
# Duplicate rows function

def drop_duplicate_rows(df):
  rows_affected = df.duplicated().sum() # Count number of rows affected
  df.drop_duplicates(inplace=True)

  return df, rows_affected

# Remove exact duplicate rows because they represent repeated records that could inflate transaction counts and sales totals
# Only rows identical across all columns are removed
# EX: rows that share a transaction_id are not automatically considered duplicates

df, rows_affected = drop_duplicate_rows(df)

log(
    'Data cleaning',
    'Drop duplicate rows',
    rows_affected
)

df.head()
```

```python
df.shape[0]
```

```python
# Negative quantity value function

def change_negative_quantities(column_name):
  rows_affected = (df[column_name] < 0).sum() # Count number of rows affected
  df[column_name] = df[column_name].abs()

  return df, rows_affected

# Convert negative quantity values to positive values using abs() so all quantities represent nonnegative units sold
# We decided to make the quantity values positive because negative quantities do not make sense
# Instead of dropping the rows, we decided to view negative quantities as user entry mistakes
df, rows_affected = change_negative_quantities('quantity')

log(
    'Data cleaning',
    'quantity: Changed negative quantity values to positive values',
    rows_affected
)
```

```python
# Parse timestamp function by converting to date time format

def parse_timestamp(column_name):
  standardized = pd.to_datetime(df[column_name], errors='coerce')
  rows_affected = df[column_name].ne(standardized).sum()
  df[column_name] = standardized

  return df, rows_affected

df, rows_affected = parse_timestamp('timestamp')

log(
    'Data cleaning',
    'timestamp: Parse timestamp function by converting to date time format',
    rows_affected
)
```
